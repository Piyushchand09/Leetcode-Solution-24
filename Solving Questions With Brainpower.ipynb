```python
# Solving Questions With Brainpower
class Solution:
    def mostPoints(self, questions: List[List[int]]) -> int:
        n = len(questions)
        best_score_from = [0 for _ in range(n + 1)]
        
        for pos in range(n - 1, -1, -1):
            points, skip = questions[pos]
            
            check_points = points
            if pos + skip < n:
                check_points += best_score_from[pos + skip + 1]
            
            if best_score_from[pos + 1] < check_points:
                best_score_from[pos] = check_points
            else:
                best_score_from[pos] = best_score_from[pos + 1]

        return best_score_from[0]
```
